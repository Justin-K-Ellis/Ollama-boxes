```python
from ollama import chat
from ollama import ChatResponse
```

```python
mdl_name = "gemma4:e2b-mlx"
sys_prompt = "You are a helpful and poetic assistant. You response to the user's requests in rhyming couplets of iambic pentameter."
```

```python
from dataclasses import dataclass


@dataclass
class ModelResponse:
   content: str
   thoughts: str | None = None


def query_model(user_query: str) -> ModelResponse:
   response: ChatResponse = chat(
      model=mdl_name,
      messages=[
         {"role": "system", "content": sys_prompt},
         {"role": "user", "content": user_query}
      ],
      think=True
   )
   if not response.message.content:
      raise Exception("something went wrong :(")
   reply = ModelResponse(
      content=response.message.content,
      thoughts=response.message.thinking
      )
   return reply
```

```python
def display_reply(reply: ModelResponse, reply_only: bool = False) -> None:
    if reply_only:
        print(reply.content)
    else:
        print("Thoughts:", reply.thoughts)
        print("")
        print("Answer:", reply.content)
```

```python
query = "Tell me about the wrath of Achilles, his fued with Agamemnon, and the fate of Hector, breaker of horses. Feel free to go on at length."

reply = query_model(query)

display_reply(reply=reply, reply_only=True)
```

```python
def talk_with_model(next_msg: str, think: bool = True) -> ModelResponse:
    response: ChatResponse = chat(
      model=mdl_name,
      messages=[
         {"role": "system", "content": sys_prompt},
         {"role": "user", "content": "My name is Justin. Please remeber that."},
         {"role": "system", "content": "I promise to remember that."},
         {"role": "user", "content": next_msg}
      ],
      think=think
   )
    if not response.message.content:
        raise Exception("something went wrong :(")
    reply = ModelResponse(
        content=response.message.content,
        thoughts=response.message.thinking
        )
    return reply
```

```python
display_reply(
    talk_with_model(
        next_msg="What is the capital of Japan. And what was my name again?",
        think=False
        )
)
```
